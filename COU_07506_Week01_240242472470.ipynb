```python
n = np.arange(-5, 11)
delta = (n == 0).astype(float)   # unit impulse
step  = (n >= 0).astype(float)   # unit step
ramp  = n * step                 # unit ramp
expo  = 0.8**n * step            # real exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title); ax.set_xlabel("n")
plt.tight_layout(); plt.show()
```

2. Why is a stem plot more appropriate than a continuous line for x(n)?

A discrete-time signal x(n) is defined only at integer values of n. There is no value between n = 2 and n = 3, for example. A stem plot shows each sample as a separate point, which matches this fact. A continuous line would wrongly suggest that the signal exists at every instant, as a continuous-time signal does.

3. For a = 1.1 the exponential grows. Which property of a system would such a signal violate if it were an impulse response?

It would violate BIBO stability. An LTI system is stable if its impulse response is absolutely summable, i.e. Σ|h(n)| < ∞. For h(n) = (1.1)ⁿu(n) the terms grow without bound, so the sum is infinite. A bounded input could therefore produce an unbounded output, and the system is unstable.

Which one repeats exactly?

cos(2πn/8) repeats exactly every 8 samples, because its period 8 is an integer. cos(2πn/7.5) does not repeat every 7.5 samples, since a sample index cannot be fractional. However, 2π/7.5 = 2π·(2/15), so its frequency f = 2/15 is rational and the sequence does repeat exactly every 15 samples (two cycles of the underlying cosine). A discrete-time sinusoid is periodic only when its frequency is a rational number.
